## 15. Object-Oriented Programming


```python
class Customer:
    def __init__(self, name, region):
        self.name = name
        self.region = region
        self.orders = []

    def add_order(self, amount):
        self.orders.append(amount)

    def total_spent(self):
        return sum(self.orders)

jane = Customer("Jane Doe", "North America")
jane.add_order(249.99)
jane.add_order(89.50)
print(jane.name, "-", jane.region)
print("Total spent:", jane.total_spent())

```

```python
class VIPCustomer(Customer):
    def __init__(self, name, region, discount_pct):
        super().__init__(name, region)
        self.discount_pct = discount_pct

    def total_spent(self):
        raw_total = sum(self.orders)
        return round(raw_total * (1 - self.discount_pct / 100), 2)

mark = VIPCustomer("Mark Lee", "Europe", 15)
mark.add_order(500)
mark.add_order(300)
print(mark.name, "total after VIP discount:", mark.total_spent())

```

## 21. Statistical Modeling



### 21.1 Hypothesis test — is Organic Search AOV different from Paid Search AOV?
An independent two-sample t-test.


```python
from scipy import stats

organic = orders[orders['channel'] == 'Organic Search']['total_amount']
paid = orders[orders['channel'] == 'Paid Search']['total_amount']

t_stat, p_value = stats.ttest_ind(organic, paid, equal_var=False)
print("Organic Search mean AOV:", round(organic.mean(), 2))
print("Paid Search mean AOV:", round(paid.mean(), 2))
print("t-statistic:", round(t_stat, 3))
print("p-value:", round(p_value, 4))

if p_value < 0.05:
    print("Result: statistically significant difference (p < 0.05)")
else:
    print("Result: no statistically significant difference (p >= 0.05)")

```

### 21.2 Correlation test — order total vs. discount

### 21.3 Simple linear regression — predicting order total from subtotal
Using `scikit-learn`'s `LinearRegression`, plus R² to see how well it fits.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

X = orders[['subtotal']]
y = orders['total_amount']

model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)
print("Coefficient (slope):", round(model.coef_[0], 4))
print("Intercept:", round(model.intercept_, 4))
print("R-squared:", round(r2_score(y, predictions), 4))

```

## 22. Final Case Study — Putting It All Together
Using a class, a loop, a function, exception handling, and pandas together in one workflow.


```python
class OrderAnalyzer:
    def __init__(self, dataframe):
        self.df = dataframe

    def total_revenue(self):
        return self.df['total_amount'].sum()

    def average_order_value(self):
        try:
            return self.df['total_amount'].mean()
        except ZeroDivisionError:
            return 0

    def best_channel(self):
        return self.df.groupby('channel')['total_amount'].sum().idxmax()

    def summary(self):
        print("Total revenue:", round(self.total_revenue(), 2))
        print("Average order value:", round(self.average_order_value(), 2))
        print("Best channel:", self.best_channel())

analyzer = OrderAnalyzer(orders)
analyzer.summary()

```
